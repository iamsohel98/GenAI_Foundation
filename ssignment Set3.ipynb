## 1. Input and Type Conversion

```
st = input('Enter a string: ')
num = input('Enter a number: ')
print(st, type(st))
print(num, type(num))
num = float(num)
print(num, type(num))
```

## 2. Conditional Grading System

```
marks = int(input('Enter marks: '))
if marks >= 90: print('A+')
elif marks >= 80: print('A')
elif marks >= 70: print('B+')
else: print('Failed')
```

## 3. Sum of Digits

```
def sum_of_digits(n):
    total=0
    while n>0:
        total+=n%10
        n//=10
    print(total)
sum_of_digits(1234)
```

## 4. Menu Driven Calculator

```
while True:
    print('1.Add 2.Subtract 3.Multiply 4.Divide 5.Exit')
    ch=int(input('Choice: '))
    if ch==5: break
    a=float(input('a: ')); b=float(input('b: '))
    if ch==1: print(a+b)
    elif ch==2: print(a-b)
    elif ch==3: print(a*b)
    elif ch==4: print(a/b)
    else: print('Invalid'); continue
```

## 5. List Comprehension

```
squares=[x**2 for x in range(1,11)]
evens=[x for x in range(1,21) if x%2==0]
chars=[c for c in 'Python']
print(squares, evens, chars)
```

## 6. Search in List

```
lst=[1,2,3,4,5]
key=3
for i in lst:
    if i==key:
        print('Found!')
        break
else:
    print('Not found')
```

## 7. Variable Arguments

```
def func(*args, **kwargs):
    print(type(args), args)
    print(type(kwargs), kwargs)
func(1,2,3,a=10,b=20)
```

## 8. Global vs Local

```
x=10
def change():
    global x
    x=20
change()
print(x)
```
