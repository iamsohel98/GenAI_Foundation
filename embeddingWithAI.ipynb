# Embedding + Claude Haiku Integration

This notebook demonstrates converting a single statement into embeddings and sending it to Claude Haiku for intelligent response generation.

## Block 1: Setup - Import Libraries and Load Model

```python
import os
from sentence_transformers import SentenceTransformer
from anthropic import Anthropic
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# Initialize the SentenceTransformer model
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
print("✓ SentenceTransformer model loaded successfully!")

# Initialize the Anthropic client
client = Anthropic(api_key=os.getenv('ANTHROPIC_API_KEY'))
print("✓ Anthropic client initialized successfully!")
```

## Block 2: Define Input Statement

```python
# Single statement to process
statement = "Machine learning is revolutionizing how we process and understand data"

print("Input Statement:")
print(f"'{statement}'")
print(f"\nLength: {len(statement)} characters")
```

## Block 3: Generate Embedding for the Statement

```python
# Generate embedding for the statement
embedding = embedding_model.encode(statement, convert_to_tensor=True)

print(f"Embedding generated successfully!")
print(f"Embedding shape: {embedding.shape}")
print(f"Total dimensions: {len(embedding)}")
print(f"\nFirst 10 dimensions of embedding:")
print(embedding[:10].tolist())
```

## Block 4: Display Embedding Statistics

```python
import numpy as np

embedding_np = embedding.cpu().numpy() if hasattr(embedding, 'cpu') else np.array(embedding)

print("=" * 60)
print("EMBEDDING STATISTICS")
print("=" * 60)
print(f"Mean value: {np.mean(embedding_np):.6f}")
print(f"Std deviation: {np.std(embedding_np):.6f}")
print(f"Min value: {np.min(embedding_np):.6f}")
print(f"Max value: {np.max(embedding_np):.6f}")
print(f"L2 Norm: {np.linalg.norm(embedding_np):.6f}")
```

## Block 5: Create Prompt with Embedding Information

```python
# Prepare the embedding data for the prompt
first_5_dims = embedding[:5].tolist()

# Create a comprehensive prompt for Claude Haiku
prompt = f"""I have converted the following statement into semantic embeddings:

Statement: "{statement}"

Embedding Details:
- Total dimensions: {embedding.shape[0]}
- First 5 dimensions: {first_5_dims}
- Embedding model: all-MiniLM-L6-v2

Based on this statement and its semantic representation, please provide:
1. A brief analysis of what this statement conveys
2. Key concepts embedded within it
3. Practical applications of this concept
4. Any related insights or considerations"""

print("Prompt prepared for Claude Haiku:")
print("=" * 60)
print(prompt)
print("=" * 60)
```

## Block 6: Send to Claude Haiku and Get Response

```python
# Send the prompt to Claude Haiku
message = client.messages.create(
    model="claude-3-5-haiku-20241022",
    max_tokens=1024,
    messages=[
        {"role": "user", "content": prompt}
    ]
)

response_text = message.content[0].text

print("✓ Response received from Claude Haiku!")
print(f"\nModel: {message.model}")
print(f"Stop reason: {message.stop_reason}")
print(f"Input tokens: {message.usage.input_tokens}")
print(f"Output tokens: {message.usage.output_tokens}")
```

## Block 7: Display Claude Haiku's Response

```python
print("\n" + "=" * 60)
print("CLAUDE HAIKU'S ANALYSIS & RESPONSE")
print("=" * 60)
print(response_text)
print("\n" + "=" * 60)
```

## Block 8: Summary and Results

```python
print("\n" + "🎯 " * 20)
print("\nEXECUTION SUMMARY")
print("\n" + "🎯 " * 20)

summary = f"""
✓ Original Statement: "{statement}"

✓ Embedding Generation:
  - Model: all-MiniLM-L6-v2
  - Dimensions: {embedding.shape[0]}
  - First 5 values: {first_5_dims}

✓ AI Response:
  - Model: Claude Haiku 3.5
  - Tokens used: {message.usage.input_tokens} (input) + {message.usage.output_tokens} (output)
  - Status: Successfully generated

✓ Integration Complete!
  The statement was successfully converted to embeddings and analyzed by Claude Haiku.
"""

print(summary)
```
