## 8. Shallow vs Deep Copy
    Create a nested list 
    • Make a shallow copy and modify an inner element 
    • Make a deep copy and modify an inner element 
    • Print all lists and explain the difference 

```python
import copy
#Shallow Copy
l1 = [[1,2],[3,4]]
shallow = copy.copy(l1)
shallow[0][0] = 100
print('Original after shallow:', l1)

#Deep Copy
l2 = [[1,2],[3,4]]
deep = copy.deepcopy(l2)
deep[0][0] = 999
print('Original after deep:', l2)
```

## 9. Logical and Bitwise Operators
    • Use and, or, and not with some conditions 
    • Use &, |, ^, ~, <<, and >> with integers 
    • Explain the output of each operation 

```python
a = 10
b = 5
print(a>5 and b<10) # True and True = True
print(a<5 or b<10) # False or True = True
print(not(a>5)) # not True = False

print(a&b) 
print(a|b)
print(a^b)
print(~a)
print(a<<1) # Shift bits left → multiply by 2
print(a>>1) # Shift bits right → divide by 2
```

## 10. Looping Through Data Structures
    • Loop through a list and print each item 
    • Loop through a dictionary and print keys, values, and items 
    • Loop through a list of tuples and unpack each tuple

```python
# list loop
for i in ['a','b','c']: 
    print(i)

# dict loop
d={'x':1,'y':2}

for k in d:
    print(k)

for v in d.values(): 
    print(v)
    
for k,v in d.items(): 
    print(k,v)

# tuple list loop
lst=[('a',1),('b',2)]

for x,y in lst: 
    print(x,y)
```
