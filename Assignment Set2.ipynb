## 6. What is the main difference between a tuple and a list?

- **List** is mutable, which means you can change, add, or remove elements.
- **Tuple** is immutable, which means you cannot change its elements after creation.

## 7. Create a function that grabs the email website domain
Example: `domainGet("user@domain.com")` should return `"domain.com"`

```
def domainGet(email):
    return email.split('@')[1]

domainGet('user@domain.com')
```

## 8. Create a function that returns True if the word 'dog' is in the input string (case-insensitive)

```
def findDog(st):
    return 'dog' in st.lower()

findDog('Is there a dog here?')
```

## 9. Create a function that counts how many times the word 'dog' occurs in a string

```
def countDog(st):
    count = 0
    for word in st.lower().split():
        if word == 'dog':
            count += 1
    return count

countDog('This dog runs faster than the other dog dude!')
```

## 10. Use lambda expressions and `filter()` to keep words that start with 's'

## 11. Final Problem - `caught_speeding(speed, is_birthday)`

```
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return 'No Ticket'
    elif 61 <= speed <= 80:
        return 'Small Ticket'
    else:
        return 'Big Ticket'

caught_speeding(81, True), caught_speeding(81, False)
```
