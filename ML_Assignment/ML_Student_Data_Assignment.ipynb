# Student Result Prediction using Decision Tree

## Objective:
- Predict student result using Decision Tree Classifier
- Visualize the decision tree
- Compute accuracy score


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
# Load dataset
df = pd.read_csv('Student_data.csv')
df.head(10)
```

```python
# Encode target variable
le = LabelEncoder()
df['Result'] = le.fit_transform(df['Result'])
df
```

```python
# Split features and target(X & Y)
X = df.drop('Result', axis=1)
y = df['Result']
```

```python
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=21)
```

```python
# Train model
model = DecisionTreeClassifier(random_state=21)
model.fit(X_train, y_train)
```

```python
# Prediction
y_pred = model.predict(X_test)
print('Predicted:', y_pred)
print('Actual:', y_test.values)
```

```python
# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy * 100:.2f}%')
```

```python
# Visualize decision tree
plt.figure(figsize=(12,8))
plot_tree(model, feature_names=X.columns, class_names=le.classes_, filled=True)
plt.show()
```
