# Exercise 1: Prompt Templates & Chaining with LangChain

**Objective**: Learn how to use LangChain's PromptTemplate and chain multiple LLM calls together.

**Learning Outcomes**:
- Create and use PromptTemplate for dynamic prompts
- Chain multiple components together
- Work with LangChain's LCEL (LangChain Expression Language)
- Use output parsers for structured responses

## Block 1: Import Libraries & Setup

```python
import os
from dotenv import load_dotenv
from langchain_anthropic import ChatAnthropic
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Load environment variables (suppress errors if .env doesn't exist)
try:
    load_dotenv(override=True)
except:
    pass

print("✓ All libraries imported successfully")
```

## Block 2: Initialize LLM

```python
# Initialize Claude model
api_key = os.environ.get("ANTHROPIC_API_KEY")
if not api_key:
    print("⚠ ANTHROPIC_API_KEY not set. Please set it in your environment.")
    print("Example: export ANTHROPIC_API_KEY='sk-your-key-here'")
else:
    llm = ChatAnthropic(
        model="global.anthropic.claude-opus-4-5-20251101-v1:0",
        api_key=api_key,
        temperature=0.7
    )
    print("✓ LLM initialized: Claude Opus")
```

## Block 3: Simple PromptTemplate Example

```python
# Create a simple prompt template
simple_prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms suitable for a 10-year-old child."
)

# Format the template
formatted_prompt = simple_prompt.format(topic="machine learning")
print("\nFormatted Prompt:")
print(formatted_prompt)
```

## Block 4: Chain - PromptTemplate + LLM

```python
if api_key:
    # Create a simple chain using LCEL (pipe operator)
    chain = simple_prompt | llm | StrOutputParser()
    
    # Invoke the chain
    result = chain.invoke({"topic": "artificial intelligence"})
    print("\nChain Result:")
    print(result)
else:
    print("Skipping - API key not configured")
```

## Block 5: Multi-Variable PromptTemplate

```python
# Create a more complex prompt with multiple variables
multi_var_prompt = PromptTemplate(
    input_variables=["topic", "difficulty", "language"],
    template="""Write a {difficulty} tutorial about {topic} in {language}.
Include:
- Key concepts
- Practical examples
- Common mistakes to avoid"""
)

# Display the template structure
print("\nMulti-Variable Prompt Template Created")
print(f"Input variables: {multi_var_prompt.input_variables}")

if api_key:
    # Create chain with multi-variable template
    multi_chain = multi_var_prompt | llm | StrOutputParser()
    
    # Invoke the chain
    result = multi_chain.invoke({
        "topic": "recursion in programming",
        "difficulty": "beginner",
        "language": "simple English"
    })
    
    print("\nMulti-Chain Result:")
    print(result)
else:
    print("Skipping execution - API key not configured")
```

## Block 6: Sequential Chain (LLM → LLM)

```python
# Define prompt templates
prompt1 = PromptTemplate(
    input_variables=["category"],
    template="Generate an interesting trivia question about {category}."
)

prompt2 = PromptTemplate(
    input_variables=["question"],
    template="Answer this trivia question: {question}\n\nProvide a concise answer with explanation."
)

print("\nSequential Chain Templates Created")
print(f"Prompt 1 inputs: {prompt1.input_variables}")
print(f"Prompt 2 inputs: {prompt2.input_variables}")

if api_key:
    parser = StrOutputParser()
    
    # Step 1: Generate question
    chain1 = prompt1 | llm | parser
    question = chain1.invoke({"category": "Science"})
    
    print("\nGenerated Question:")
    print(question)
    
    # Step 2: Answer the question
    chain2 = prompt2 | llm | parser
    answer = chain2.invoke({"question": question})
    
    print("\nGenerated Answer:")
    print(answer)
else:
    print("Skipping execution - API key not configured")
```
